# Script for online analysis and visualization of data during the IVU transverse impedance measurement Machine Study

The goal of this script is to provide functions and tools to:

- Read the data from the acquisition script;
- Visualize the data;
- Perform fits to extract the current-normalized orbit deviations;
- Calculate and display useful statistics about the data and the fit

## Imports

```python
import os
import numpy as np
import matplotlib.pyplot as mplt

from mathphys.functions import load  # Loads data (and params) from .pickle file
```

## Handling the data

Data description:

Each data file from the acquisition script will be a dictionary with keys:

-  buc1_xpos $\,$ | $\,$ buc1_ypos $\,$ | $\,$ buc1_sum $\:$ ~ $\:$ _all (160, n_acq, n_turns)_

-  buc2_xpos $\,$ | $\,$ buc2_ypos $\,$ | $\,$ buc2_sum $\:$ ~ $\:$ _all (160, n_acq, n_turns)_


- info (dict):

    - current_mA
    - sofb_orb_x_µm
    - sofb_orb_y_µm
    - corr_forces
    - timestamp
.
- config (dict):

    - gap_mm
    - y0_mm
    - delta_q_target_nC
.
- derived (dict):

    - buc1_current_mA
    - buc2_current_mA
    - delta_current_mA
    - buc1_charge_nC
    - buc2_charge_nC
    - delta_charge_nC
