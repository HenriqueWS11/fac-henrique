# Synthetic data for the IVU transverse impedance measurement

This notebook generates synthetic turn-by-turn BPM data with the same structure expected from the acquisition script. The goal is not to simulate the full machine perfectly, but to create controlled data where the true IVU impedance signal is known. This allows us to test the analysis and visualization tools before the real machine study.

The synthetic model is based on the current-normalized orbit-difference equation:

\begin{equation}
    \frac{\Delta \mathbf u}{\Delta q} = \mathbf c^u + y_0\,\mathbf M_{\mathrm{IVU}}^u \frac{k_\perp^u}{E/e} + \boldsymbol\eta^u
\end{equation}

where $u=x,y$. Each generated file corresponds to one measured configuration: fixed IVU gap, fixed bump value $y_0$, and fixed bunch-charge difference $\Delta q$.
