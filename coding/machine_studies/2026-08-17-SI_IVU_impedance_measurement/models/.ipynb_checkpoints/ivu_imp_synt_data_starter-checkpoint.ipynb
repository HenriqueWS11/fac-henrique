# Synthetic data generation for the IVU transverse impedance measurement

This notebook generates **synthetic raw data files** with the same basic structure expected from the IVU transverse impedance machine-study acquisition script.

The goal is not to perfectly reproduce the machine, but to create controlled data where we know the true impedance signal. This allows us to test the online analysis notebook before the measurement:

- load several configuration files;
- compute bunch-by-bunch orbit differences;
- normalize by bunch-charge difference;
- fit the expected IVU response pattern;
- visualize residuals, fit quality, and scan trends.

Each synthetic file corresponds to **one measured configuration**, i.e. fixed:

\[
(g,\; y_0,\; \Delta q_{\mathrm{target}})
\]

where:

- \(g\) is the IVU gap;
- \(y_0\) is the vertical bump amplitude at the IVU;
- \(\Delta q = q_1-q_2\) is the charge difference between the two bunches.

Inside each file, the turn-by-turn data are stored as six arrays with shape:

\[
(160,\; N_{\mathrm{turns,total}})
\]

for:

\[
\texttt{buc1\_xpos},\;\texttt{buc2\_xpos},\;\texttt{buc1\_ypos},\;\texttt{buc2\_ypos},\;\texttt{buc1\_sum},\;\texttt{buc2\_sum}.
\]

## 1. Synthetic model

For each plane \(u\in\{x,y\}\), BPM \(i\), and turn \(n\), define the bunch-by-bunch orbit difference:

\[
\Delta u_{i,n} = u_{1,i,n}-u_{2,i,n}.
\]

The charge difference is:

\[
\Delta q = q_1-q_2.
\]

The normalized orbit difference is:

\[
z_{i,n}^{u}=\frac{\Delta u_{i,n}}{\Delta q}.
\]

For the synthetic data, we use the model:

\[
\boxed{
z_{i,n}^{u}
=
 c_i^{u}(g)
+
y_0\,M_{\mathrm{IVU},i}^{u}(y_0,g)\,\beta^{u}(g)
+
\eta_{i,n}^{u}
}
\]

where:

- \(c_i^{u}(g)\) is a repeatable charge-dependent background pattern, independent of \(y_0\);
- \(M_{\mathrm{IVU},i}^{u}\) is the orbit-response column at BPM \(i\) for a unit kick at the IVU;
- \(\beta^{u}(g)\) is the fitted scalar response amplitude;
- \(\eta_{i,n}^{u}\) is the remaining turn-by-turn fluctuation.

In physical units, if all quantities are in SI units,

\[
\boxed{
\beta^{u}(g)=\frac{k_{\perp}^{u}(g)}{E/e}
}
\]

so that the fit gives:

\[
\boxed{
k_{\perp}^{u}(g)=(E/e)\,\beta^{u}(g).
}
\]

In this notebook, for convenience, we use practical units:

\[
M_{\mathrm{IVU}}: \mu\mathrm{m}/\mu\mathrm{rad},
\qquad
\beta: \mu\mathrm{rad}/(\mathrm{nC}\,\mathrm{mm}),
\qquad
y_0: \mathrm{mm}.
\]

Then:

\[
y_0\,M_{\mathrm{IVU}}\,\beta
\quad\text{has units}\quad
\mu\mathrm{m}/\mathrm{nC}.
\]

## 2. Imports

```python
from pathlib import Path
from datetime import datetime
import pickle

import numpy as np
import pandas as pd
import matplotlib.pyplot as mplt
```

## 3. Global settings

These settings define the synthetic measurement campaign. They can be changed later to mimic the real scan.

```python
# Reproducibility
RNG_SEED = 20260803
rng = np.random.default_rng(RNG_SEED)

# Basic dimensions
N_BPM = 160
N_ACQ = 20                 # Stored as metadata. The raw arrays are concatenated over acquisitions.
N_TURNS_PER_ACQ = 10
N_TURNS_TOTAL = N_ACQ * N_TURNS_PER_ACQ

# Approximate revolution frequency. Check/update with the actual machine value.
f0_kHz = 578.0

# Synthetic scan values
GAPS_MM = [4.0, 6.0, 8.0]
Y0_VALUES_MM = [-1.0, 0.0, 1.0, 2.0]
DELTA_Q_TARGETS_NC = [0.6, 1.0, 1.4]

# Low-charge bunch. The high-charge bunch is q_low + delta_q.
Q_LOW_NC = 0.25

# Output folder
PROJECT_ROOT = Path.cwd()
SYNTHETIC_FOLDER = PROJECT_ROOT / "data" / "synthetic"
SYNTHETIC_FOLDER.mkdir(parents=True, exist_ok=True)

print("Synthetic folder:", SYNTHETIC_FOLDER)
print("Total configurations:", len(GAPS_MM) * len(Y0_VALUES_MM) * len(DELTA_Q_TARGETS_NC))
```

## 4. Helper functions: filenames, currents, and response columns

```python
def fmt_float_for_filename(value: float) -> str:
    """Convert a float to a filename-safe string."""
    txt = f"{value:.3g}".replace("-", "m").replace(".", "_")
    return txt


def charge_nC_to_current_mA(q_nC: float, f0_kHz: float) -> float:
    """Convert single-bunch charge in nC to single-bunch current in mA.

    q [nC] * f0 [kHz] * 1e-3 = I [mA]
    """
    return q_nC * f0_kHz * 1e-3


def current_mA_to_charge_nC(current_mA: float, f0_kHz: float) -> float:
    """Convert single-bunch current in mA to single-bunch charge in nC."""
    return current_mA / (f0_kHz * 1e-3)


def make_response_column(n_bpm: int, amplitude_um_per_urad: float, phase: float = 0.0) -> np.ndarray:
    """Create a smooth synthetic orbit-response column for a unit IVU kick.

    In real analysis, this vector should come from the optics model by applying
    a small virtual kick at the IVU and reading the closed orbit at BPMs.
    """
    s = np.linspace(0.0, 2.0 * np.pi, n_bpm, endpoint=False)
    column = (
        np.cos(s + phase)
        + 0.35 * np.cos(2.0 * s - 0.4 + phase)
        + 0.15 * np.sin(3.0 * s + 0.7)
    )
    column -= np.mean(column)
    column /= np.std(column)
    return amplitude_um_per_urad * column


def make_background_vector(n_bpm: int, amplitude_um_per_nC: float, phase: float = 0.0) -> np.ndarray:
    """Create a repeatable charge-dependent background pattern independent of y0."""
    s = np.linspace(0.0, 2.0 * np.pi, n_bpm, endpoint=False)
    c = (
        np.sin(1.0 * s + phase)
        + 0.25 * np.cos(4.0 * s - 0.2)
    )
    c -= np.mean(c)
    c /= np.std(c)
    return amplitude_um_per_nC * c


def make_bump_orbit_shape(n_bpm: int, y0_mm: float, plane: str = "y") -> np.ndarray:
    """Synthetic common closed-orbit bump seen by both bunches.

    This is not the wake signal. It cancels when buc1 - buc2 is computed,
    except for possible BPM/systematic effects.
    """
    s = np.linspace(0.0, 2.0 * np.pi, n_bpm, endpoint=False)
    if plane == "y":
        shape = np.cos(s - 0.2) + 0.2 * np.cos(2.0 * s + 0.8)
        amplitude_um = 120.0 * y0_mm
    else:
        shape = 0.25 * np.sin(s + 0.4)
        amplitude_um = 20.0 * y0_mm
    shape -= np.mean(shape)
    shape /= np.max(np.abs(shape))
    return amplitude_um * shape
```

## 5. Synthetic truth model

Here we define the hidden truth used to generate the data.

The vertical signal is intentionally stronger because the planned measurement is a vertical bump and vertical transverse impedance measurement. The horizontal plane is generated as a weaker control channel.

```python
def make_truth_for_gap(gap_mm: float) -> dict:
    """Return the synthetic truth parameters for a given IVU gap."""
    # Smaller gap -> stronger impedance signal.
    # These values are synthetic and only chosen to create visible signals.
    gap_scale = (4.0 / gap_mm) ** 2

    truth = {
        "M_IVU_y_um_per_urad": make_response_column(N_BPM, amplitude_um_per_urad=0.80, phase=0.15 * gap_mm),
        "M_IVU_x_um_per_urad": make_response_column(N_BPM, amplitude_um_per_urad=0.12, phase=0.10 * gap_mm + 1.0),
        "c_y_um_per_nC": make_background_vector(N_BPM, amplitude_um_per_nC=0.35, phase=0.20 * gap_mm),
        "c_x_um_per_nC": make_background_vector(N_BPM, amplitude_um_per_nC=0.20, phase=0.15 * gap_mm + 0.5),
        # beta = dtheta / (delta_q * y0)
        # Units used here: microrad / (nC mm)
        "beta_y_urad_per_nC_mm": 0.90 * gap_scale,
        "beta_x_urad_per_nC_mm": 0.08 * gap_scale,
    }
    return truth


TRUTH_BY_GAP = {gap: make_truth_for_gap(gap) for gap in GAPS_MM}

# Quick look at the synthetic response column for one gap.
gap0 = GAPS_MM[0]
mplt.figure()
mplt.plot(TRUTH_BY_GAP[gap0]["M_IVU_y_um_per_urad"], label="Vertical")
mplt.plot(TRUTH_BY_GAP[gap0]["M_IVU_x_um_per_urad"], label="Horizontal")
mplt.xlabel("BPM index")
mplt.ylabel(r"$M_{IVU}$ [$\mu$m/$\mu$rad]")
mplt.title(f"Synthetic IVU response columns, gap = {gap0} mm")
mplt.legend()
mplt.grid(True, alpha=0.3)
mplt.show()
```

## 6. Generate one configuration

For one configuration, the normalized model is:

\[
\frac{\Delta u_{i,n}}{\Delta q}
=
 c_i^u(g)
+
y_0\,M_{\mathrm{IVU},i}^{u}(g)\,\beta^u(g)
+
\eta_{i,n}^{u}.
\]

Then the raw bunch difference is:

\[
\Delta u_{i,n}
=
\Delta q\left[
 c_i^u(g)
+
y_0\,M_{\mathrm{IVU},i}^{u}(g)\,\beta^u(g)
+
\eta_{i,n}^{u}
\right].
\]

The individual bunch positions are built as:

\[
u_{1,i,n}=u_{\mathrm{common},i,n}+\frac{1}{2}\Delta u_{i,n}+\xi_{1,i,n},
\]

\[
u_{2,i,n}=u_{\mathrm{common},i,n}-\frac{1}{2}\Delta u_{i,n}+\xi_{2,i,n}.
\]

The common orbit cancels in the bunch difference, while \(\xi_1\) and \(\xi_2\) represent BPM readout noise.

```python
def estimate_currents_from_sums(buc1_sum: np.ndarray, buc2_sum: np.ndarray, total_current_mA: float, f0_kHz: float) -> dict:
    """Estimate bunch currents and charges from BPM sum signals."""
    frac1 = np.nanmedian(buc1_sum / (buc1_sum + buc2_sum))
    frac2 = 1.0 - frac1

    buc1_current_mA = frac1 * total_current_mA
    buc2_current_mA = frac2 * total_current_mA
    delta_current_mA = buc1_current_mA - buc2_current_mA

    buc1_charge_nC = current_mA_to_charge_nC(buc1_current_mA, f0_kHz)
    buc2_charge_nC = current_mA_to_charge_nC(buc2_current_mA, f0_kHz)
    delta_charge_nC = buc1_charge_nC - buc2_charge_nC

    return {
        "buc1_current_mA": buc1_current_mA,
        "buc2_current_mA": buc2_current_mA,
        "delta_current_mA": delta_current_mA,
        "buc1_charge_nC": buc1_charge_nC,
        "buc2_charge_nC": buc2_charge_nC,
        "delta_charge_nC": delta_charge_nC,
    }


def generate_sum_signals(q1_nC: float, q2_nC: float, n_bpm: int, n_turns: int, rng: np.random.Generator) -> tuple[np.ndarray, np.ndarray]:
    """Generate synthetic BPM sum signals proportional to bunch charge."""
    nominal_sum = 1.0e5
    bpm_gain = 1.0 + rng.normal(0.0, 0.015, size=(n_bpm, 1))
    turn_gain = 1.0 + rng.normal(0.0, 0.003, size=(1, n_turns))
    noise1 = 1.0 + rng.normal(0.0, 0.004, size=(n_bpm, n_turns))
    noise2 = 1.0 + rng.normal(0.0, 0.004, size=(n_bpm, n_turns))

    buc1_sum = nominal_sum * q1_nC * bpm_gain * turn_gain * noise1
    buc2_sum = nominal_sum * q2_nC * bpm_gain * turn_gain * noise2
    return buc1_sum, buc2_sum


def generate_configuration_data(
    gap_mm: float,
    y0_mm: float,
    delta_q_target_nC: float,
    rng: np.random.Generator,
    n_bpm: int = N_BPM,
    n_turns_total: int = N_TURNS_TOTAL,
    n_acq: int = N_ACQ,
    n_turns_per_acq: int = N_TURNS_PER_ACQ,
    q_low_nC: float = Q_LOW_NC,
    f0_kHz: float = f0_kHz,
    eta_y_um_per_nC: float = 0.08,
    eta_x_um_per_nC: float = 0.06,
    bpm_readout_noise_um: float = 0.04,
) -> dict:
    """Generate one synthetic configuration dictionary."""
    truth = TRUTH_BY_GAP[gap_mm]

    # Actual charge difference is close to the target, but not exactly equal.
    delta_q_actual_nC = delta_q_target_nC * (1.0 + rng.normal(0.0, 0.015))
    q2_nC = q_low_nC
    q1_nC = q2_nC + delta_q_actual_nC

    buc1_current_mA_true = charge_nC_to_current_mA(q1_nC, f0_kHz)
    buc2_current_mA_true = charge_nC_to_current_mA(q2_nC, f0_kHz)
    total_current_mA = buc1_current_mA_true + buc2_current_mA_true

    # Response columns and background vectors.
    M_y = truth["M_IVU_y_um_per_urad"]
    M_x = truth["M_IVU_x_um_per_urad"]
    c_y = truth["c_y_um_per_nC"]
    c_x = truth["c_x_um_per_nC"]
    beta_y = truth["beta_y_urad_per_nC_mm"]
    beta_x = truth["beta_x_urad_per_nC_mm"]

    # Normalized turn-by-turn residuals eta.
    eta_y = rng.normal(0.0, eta_y_um_per_nC, size=(n_bpm, n_turns_total))
    eta_x = rng.normal(0.0, eta_x_um_per_nC, size=(n_bpm, n_turns_total))

    # Normalized bunch differences z = Delta position / Delta q.
    z_y = c_y[:, None] + y0_mm * M_y[:, None] * beta_y + eta_y
    z_x = c_x[:, None] + y0_mm * M_x[:, None] * beta_x + eta_x

    # Raw bunch differences in micrometers.
    delta_y_um = delta_q_actual_nC * z_y
    delta_x_um = delta_q_actual_nC * z_x

    # Common orbit seen by both bunches.
    common_y = make_bump_orbit_shape(n_bpm, y0_mm, plane="y")[:, None]
    common_x = make_bump_orbit_shape(n_bpm, y0_mm, plane="x")[:, None]
    common_turn_jitter_y = rng.normal(0.0, 0.03, size=(1, n_turns_total))
    common_turn_jitter_x = rng.normal(0.0, 0.03, size=(1, n_turns_total))

    # Individual bunch readout noise.
    noise1_y = rng.normal(0.0, bpm_readout_noise_um, size=(n_bpm, n_turns_total))
    noise2_y = rng.normal(0.0, bpm_readout_noise_um, size=(n_bpm, n_turns_total))
    noise1_x = rng.normal(0.0, bpm_readout_noise_um, size=(n_bpm, n_turns_total))
    noise2_x = rng.normal(0.0, bpm_readout_noise_um, size=(n_bpm, n_turns_total))

    buc1_ypos = common_y + common_turn_jitter_y + 0.5 * delta_y_um + noise1_y
    buc2_ypos = common_y + common_turn_jitter_y - 0.5 * delta_y_um + noise2_y
    buc1_xpos = common_x + common_turn_jitter_x + 0.5 * delta_x_um + noise1_x
    buc2_xpos = common_x + common_turn_jitter_x - 0.5 * delta_x_um + noise2_x

    # Sum signals and derived current estimates.
    buc1_sum, buc2_sum = generate_sum_signals(q1_nC, q2_nC, n_bpm, n_turns_total, rng)
    derived = estimate_currents_from_sums(buc1_sum, buc2_sum, total_current_mA, f0_kHz)

    # Corrector forces are synthetic placeholders.
    corr_forces = np.array([-1.0, 1.0, 1.0, -1.0]) * y0_mm

    data = {
        "buc1_xpos": buc1_xpos.astype(np.float32),
        "buc2_xpos": buc2_xpos.astype(np.float32),
        "buc1_ypos": buc1_ypos.astype(np.float32),
        "buc2_ypos": buc2_ypos.astype(np.float32),
        "buc1_sum": buc1_sum.astype(np.float32),
        "buc2_sum": buc2_sum.astype(np.float32),
        "info": {
            "total_current_mA": float(total_current_mA),
            "sofb_orbit_x_um": np.mean(common_x, axis=1).astype(np.float32),
            "sofb_orbit_y_um": np.mean(common_y, axis=1).astype(np.float32),
            "corr_forces": corr_forces.astype(np.float32),
            "timestamp": datetime.now().isoformat(timespec="seconds"),
        },
        "config": {
            "gap_mm": float(gap_mm),
            "y0_mm": float(y0_mm),
            "delta_q_target_nC": float(delta_q_target_nC),
            "n_bpm": int(n_bpm),
            "n_acq": int(n_acq),
            "n_turns_per_acq": int(n_turns_per_acq),
            "n_turns_total": int(n_turns_total),
            "f0_kHz": float(f0_kHz),
        },
        "derived": derived,
        "synthetic_truth": {
            "q1_true_nC": float(q1_nC),
            "q2_true_nC": float(q2_nC),
            "delta_q_true_nC": float(delta_q_actual_nC),
            "beta_y_urad_per_nC_mm": float(beta_y),
            "beta_x_urad_per_nC_mm": float(beta_x),
            "M_IVU_y_um_per_urad": M_y.astype(np.float32),
            "M_IVU_x_um_per_urad": M_x.astype(np.float32),
            "c_y_um_per_nC": c_y.astype(np.float32),
            "c_x_um_per_nC": c_x.astype(np.float32),
        },
    }
    return data
```

## 7. Save and load utilities

The real notebook currently plans to load `.pickle` files. Here we save the synthetic configuration dictionaries in the same style.

```python
def save_pickle(obj: dict, file_path: Path) -> None:
    file_path.parent.mkdir(parents=True, exist_ok=True)
    with open(file_path, "wb") as f:
        pickle.dump(obj, f, protocol=pickle.HIGHEST_PROTOCOL)


def load_pickle(file_path: Path) -> dict:
    with open(file_path, "rb") as f:
        return pickle.load(f)


def synthetic_filename(gap_mm: float, delta_q_target_nC: float, y0_mm: float) -> str:
    gap_txt = fmt_float_for_filename(gap_mm)
    dq_txt = fmt_float_for_filename(delta_q_target_nC)
    y0_txt = fmt_float_for_filename(y0_mm)
    return f"imp_meas_gap_{gap_txt}_Deltaq_{dq_txt}_y0_{y0_txt}.pickle"
```

## 8. Generate the full synthetic scan

This cell creates one file per configuration and a `manifest.csv` describing the scan.

```python
manifest_rows = []

for gap_mm in GAPS_MM:
    for delta_q_target_nC in DELTA_Q_TARGETS_NC:
        for y0_mm in Y0_VALUES_MM:
            data = generate_configuration_data(
                gap_mm=gap_mm,
                y0_mm=y0_mm,
                delta_q_target_nC=delta_q_target_nC,
                rng=rng,
            )
            filename = synthetic_filename(gap_mm, delta_q_target_nC, y0_mm)
            file_path = SYNTHETIC_FOLDER / filename
            save_pickle(data, file_path)

            manifest_rows.append({
                "filename": filename,
                "file_path": str(file_path),
                "gap_mm": gap_mm,
                "y0_mm": y0_mm,
                "delta_q_target_nC": delta_q_target_nC,
                "delta_q_measured_nC": data["derived"]["delta_charge_nC"],
                "n_bpm": data["config"]["n_bpm"],
                "n_turns_total": data["config"]["n_turns_total"],
                "timestamp": data["info"]["timestamp"],
            })

manifest = pd.DataFrame(manifest_rows)
manifest_path = SYNTHETIC_FOLDER / "manifest.csv"
manifest.to_csv(manifest_path, index=False)

print(f"Saved {len(manifest)} synthetic configuration files.")
print("Manifest:", manifest_path)
manifest.head()
```

## 9. Quick sanity check of one file

Here we load one file and verify the expected structure.

```python
example_path = Path(manifest.iloc[0]["file_path"])
example = load_pickle(example_path)

print("Top-level keys:", list(example.keys()))
print("Raw array shape:", example["buc1_ypos"].shape)
print("Config:", example["config"])
print("Derived:", example["derived"])
```

## 10. First reduction: normalized orbit difference

For one configuration and plane, compute:

\[
\mathbf z_n = \frac{\mathbf u_{1,n}-\mathbf u_{2,n}}{\Delta q}.
\]

Then calculate the mean and standard error over turns.

```python
def get_plane_positions(data: dict, plane: str) -> tuple[np.ndarray, np.ndarray]:
    if plane == "x":
        return data["buc1_xpos"], data["buc2_xpos"]
    if plane == "y":
        return data["buc1_ypos"], data["buc2_ypos"]
    raise ValueError("plane must be 'x' or 'y'")


def normalized_orbit_difference(data: dict, plane: str = "y") -> np.ndarray:
    """Return z = (buc1 - buc2) / delta_q with shape (BPM, turns)."""
    buc1, buc2 = get_plane_positions(data, plane)
    delta_q = data["derived"]["delta_charge_nC"]
    return (buc1 - buc2) / delta_q


def reduce_configuration(data: dict, plane: str = "y") -> dict:
    z = normalized_orbit_difference(data, plane=plane)
    n_turns = z.shape[1]
    return {
        "z_mean_um_per_nC": np.mean(z, axis=1),
        "z_std_um_per_nC": np.std(z, axis=1, ddof=1),
        "z_sem_um_per_nC": np.std(z, axis=1, ddof=1) / np.sqrt(n_turns),
        "n_turns": n_turns,
    }


red_y = reduce_configuration(example, plane="y")
red_x = reduce_configuration(example, plane="x")

mplt.figure()
mplt.plot(red_y["z_mean_um_per_nC"], label="Vertical")
mplt.plot(red_x["z_mean_um_per_nC"], label="Horizontal")
mplt.xlabel("BPM index")
mplt.ylabel(r"$\langle \Delta u / \Delta q \rangle$ [$\mu$m/nC]")
mplt.title("Normalized bunch-orbit difference for one synthetic configuration")
mplt.legend()
mplt.grid(True, alpha=0.3)
mplt.show()
```

## 11. Direct fit with background vector

For one fixed gap, the normalized model is:

\[
\mathbf z_m
=
\mathbf c
+
y_{0,m}\mathbf M_{\mathrm{IVU}}(y_{0,m},g)\beta(g)
+
\boldsymbol\eta_m.
\]

If we fit all configurations of one gap together, \(\mathbf c\) is a BPM-dependent intercept vector and \(\beta\) is one scalar.

With equal weights, subtracting the BPM-wise mean removes \(\mathbf c\). The fitted scalar is:

\[
\boxed{
\hat\beta
=
\frac{\sum_{m,i}s'_{i,m}z'_{i,m}}
{\sum_{m,i}\left(s'_{i,m}\right)^2}
}
\]

with:

\[
\mathbf s_m=y_{0,m}\mathbf M_{\mathrm{IVU}}(y_{0,m},g).
\]

```python
def fit_beta_for_gap(manifest: pd.DataFrame, gap_mm: float, plane: str = "y") -> dict:
    """Fit beta for one gap using all charge and y0 configurations.

    This uses the synthetic-truth M_IVU column stored in each file. In real data,
    replace this by the model/simulated M_IVU column.
    """
    rows = manifest[manifest["gap_mm"] == gap_mm].copy()
    z_list = []
    s_list = []
    y0_list = []
    dq_list = []

    for _, row in rows.iterrows():
        data = load_pickle(Path(row["file_path"]))
        red = reduce_configuration(data, plane=plane)
        z = red["z_mean_um_per_nC"]
        y0 = data["config"]["y0_mm"]
        if plane == "y":
            M = data["synthetic_truth"]["M_IVU_y_um_per_urad"]
        else:
            M = data["synthetic_truth"]["M_IVU_x_um_per_urad"]
        s = y0 * M

        z_list.append(z)
        s_list.append(s)
        y0_list.append(y0)
        dq_list.append(data["derived"]["delta_charge_nC"])

    Z = np.stack(z_list, axis=0)  # (n_config_gap, n_bpm)
    S = np.stack(s_list, axis=0)  # (n_config_gap, n_bpm)

    # Remove the BPM-dependent intercept vector c.
    Zc = Z - np.mean(Z, axis=0, keepdims=True)
    Sc = S - np.mean(S, axis=0, keepdims=True)

    beta_hat = np.sum(Sc * Zc) / np.sum(Sc * Sc)
    residual = Zc - beta_hat * Sc
    rms_residual = np.sqrt(np.mean(residual**2))

    return {
        "gap_mm": gap_mm,
        "plane": plane,
        "beta_hat_urad_per_nC_mm": beta_hat,
        "rms_residual_um_per_nC": rms_residual,
        "n_configs_used": len(rows),
        "Z": Z,
        "S": S,
        "y0_values": np.array(y0_list),
        "delta_q_values": np.array(dq_list),
    }


fit_y = fit_beta_for_gap(manifest, gap_mm=GAPS_MM[0], plane="y")
fit_x = fit_beta_for_gap(manifest, gap_mm=GAPS_MM[0], plane="x")

print("Vertical fit:", fit_y)
print("Horizontal fit:", fit_x)
print("True beta y:", TRUTH_BY_GAP[GAPS_MM[0]]["beta_y_urad_per_nC_mm"])
print("True beta x:", TRUTH_BY_GAP[GAPS_MM[0]]["beta_x_urad_per_nC_mm"])
```

## 12. Visual check: scan collapse versus bump

A useful online diagnostic is to project the normalized BPM-vector data onto the expected IVU response direction, or simply look at representative BPMs.

For an ideal linear signal, after subtracting the fitted background, the data should scale linearly with \(y_0\).

```python
def plot_representative_bpms(manifest: pd.DataFrame, gap_mm: float, plane: str = "y", bpm_indices=(10, 40, 80, 120)):
    rows = manifest[manifest["gap_mm"] == gap_mm].copy()

    mplt.figure()
    for bpm in bpm_indices:
        xs = []
        ys = []
        for _, row in rows.iterrows():
            data = load_pickle(Path(row["file_path"]))
            z = reduce_configuration(data, plane=plane)["z_mean_um_per_nC"]
            xs.append(data["config"]["y0_mm"])
            ys.append(z[bpm])
        mplt.scatter(xs, ys, label=f"BPM {bpm}")

    mplt.xlabel(r"$y_0$ [mm]")
    mplt.ylabel(r"$\langle \Delta u / \Delta q \rangle$ [$\mu$m/nC]")
    mplt.title(f"Representative BPMs, plane={plane}, gap={gap_mm} mm")
    mplt.legend()
    mplt.grid(True, alpha=0.3)
    mplt.show()


plot_representative_bpms(manifest, gap_mm=GAPS_MM[0], plane="y")
```

This synthetic notebook now produces raw-like data files. The analysis notebook can use these files exactly as it will later use the real acquisition files.

Useful next improvements:

\mathbf z_{\mathrm{sub}}(y_0)

\mathbf z(y_0)-\mathbf z(0).
\]

3. Add bad BPMs and outlier turns to test robustness.
4. Add a small bump-leakage source to test whether residual plots can detect contamination from another impedance source.
